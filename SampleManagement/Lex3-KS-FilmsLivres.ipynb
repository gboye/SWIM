```python
import pandas as pd
from scipy import stats
```

```python
nomLexiqueOrg="/Volumes/BroadExt/Dropbox/Travail/Recherche/BoBoHe/datasets/vlexique/Lexique3.txt"
```

```python
dfL=pd.read_csv(nomLexiqueOrg,sep="\t")
```

```python
df=dfL[['3_lemme', '4_cgram', '5_genre', '6_nombre', '9_freqfilms2', '10_freqlivres','11_infover']]
df.columns=["lemme","cat","genre","nombre","freqF","freqL","ver"]
dfV=df.loc[df.cat.isin(["AUX","VER"])]
dfV.head()
```

```python
dfV['freqF'] = dfV['freqF'].fillna(0)
dfV['relF'] = dfV['freqF'] / dfV['freqF'].sum()
print("total Films",dfV['freqF'].sum())
dfV['freqL'] = dfV['freqL'].fillna(0)
dfV['relL'] = dfV['freqL'] / dfV['freqL'].sum()
print("total Livres",dfV['freqL'].sum())
statistic,p_value=stats.ks_2samp(dfV['relF'], dfV['relL'])
print(f"Statistique KS : {statistic:.2%}")
print(f"p-value : {p_value:.8f}\n")  
```

```python
seuil=10
cond=(dfV.freqF>seuil)&(dfV.freqL>seuil)
statistic,p_value=stats.ks_2samp(dfV.loc[cond]['relF'], dfV.loc[cond]['relL'])
print(f"Statistique KS : {statistic:.2%}")
print(f"p-value : {p_value:.8f}\n")  
dfV.loc[cond]
```

```python
df['freqF'] = df['freqF'].fillna(0)
df['relF'] = df['freqF'] / df['freqF'].sum()
print("total Films",df['freqF'].sum())
df['freqL'] = df['freqL'].fillna(0)
df['relL'] = df['freqL'] / df['freqL'].sum()
print("total Livres",df['freqL'].sum())
statistic,p_value=stats.ks_2samp(df['relF'], df['relL'])
print(f"Statistique KS : {statistic:.2%}")
print(f"p-value : {p_value:.8f}\n")  
```
